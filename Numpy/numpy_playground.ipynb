```python
matrix_of_sevens = np.ones((5,5)) * 7
matrix_of_sevens
```

```python
vector = np.arange(5).reshape(5,-1)
vector
```

```python
print(matrix_of_sevens * vector)

print(vector * matrix_of_sevens)
```

```python
# Broadcasting rules
# 1. Padding (adding ones to the left of the shape of the smaller ndarray)
#  when (3,7) matrix added to (7) vector , (7) vector is converted to (1,7) matrix

# 2. Comparison (broadcasting happens if each pair of entries are equal or one of them are one)

# 3. Broadcast can be forced if  vector and matrix share same number of rows or column and one of dimension of vector is 1
# ie if matrix is (x,y) both vector (x,1) and (1,y) can be broadcasted to the matrix
```

```python
matrix = np.zeros((3,7)) # (3,7)
vector = np.arange(1,8) # (7)
print(matrix)
print(vector)
```

```python
matrix + vector
```
